```python
a1 = np.array([[1,1], 
               [2,2]])
a2 = np.array([[3,3],
              [4,4]])
print(f'a1:\n{a1}')
print(f'a2:\n{a2}')
```

```python
# Stack the arrays vertically
vert_stack = np.vstack((a1, a2))
print(vert_stack)
```

```python
# Stack the arrays horizontally
horz_stack = np.hstack((a1, a2))
print(horz_stack)
```
